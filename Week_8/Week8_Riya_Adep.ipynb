## Intelligent Query Routing Agent

This project implements a simple Intelligent Query Routing Agent capable of directing user queries to specialized tools based on the query content.

The agent supports three types of requests:

- Mathematical calculations
- Keyword extraction
- General conversational queries

The system returns structured JSON responses containing both the response type and the generated result, providing a simple demonstration of rule-based AI agent routing.

## Step 2: Build the Calculator Tool

The calculator tool evaluates mathematical expressions provided by the user.

```python
# Calculator Tool

def calculator(expression):

    try:
        result = eval(expression)

        return result

    except Exception as e:

        return f"Error: {e}"
```

## Step 3: Build the Keyword Extraction Tool

The keyword extraction tool identifies the most meaningful words from a given text.

The input sentence is converted to lowercase, punctuation is removed, and common English stop words are filtered out. The remaining words are returned as extracted keywords, providing a simple representation of the important terms present in the text.

```python
# Keyword Extraction Tool

def extract_keywords(text):

    # Common English stop words
    stop_words = {
        "the", "is", "a", "an", "and", "or", "of", "to",
        "in", "on", "for", "with", "at", "by", "from",
        "this", "that", "it", "are", "was", "were", "be",
        "as", "has", "have", "had"
    }

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    words = re.findall(r"\b[a-zA-Z]+\b", text)

    # Remove stop words
    keywords = [word for word in words if word not in stop_words]

    return keywords
```

## Step 5: Build the Intelligent Query Routing Agent

The intelligent query routing agent acts as the central controller of the system. It receives a user's query, identifies the intended task using conditional pattern matching, and routes the request to the appropriate processing tool.

The routing logic follows these rules:

- Queries containing **"calculate"** are sent to the calculator tool.
- Queries containing **"keywords"** are sent to the keyword extraction tool.
- All other queries are handled by a general fallback response.
- Invalid inputs or processing errors are returned as structured error responses.

Each response is formatted as a JSON object containing two fields:

- **type** – identifies the category of the response.
- **result** – contains the generated output.

```python
# Intelligent Query Routing Agent

def agent(query):

    try:

        # Validate input
        if not isinstance(query, str) or not query.strip():
            return {
                "type": "error",
                "result": "Invalid input. Please enter a valid query."
            }

        query_lower = query.lower()

        # ----------------------------
        # Route to Calculator
        # ----------------------------
        if "calculate" in query_lower:

            expression = query_lower.replace("calculate", "").strip()

            result = calculator(expression)

            return {
                "type": "calculation",
                "result": result
            }

        # ----------------------------
        # Route to Keyword Extractor
        # ----------------------------
        elif "keywords" in query_lower:

            text = (
              query_lower
              .replace("extract keywords from", "")
              .replace("keywords", "")
              .strip()
             )

            result = extract_keywords(text)

            return {
                "type": "keywords",
                "result": result
            }

        # ----------------------------
        # General Fallback
        # ----------------------------
        else:

            return {
                "type": "general",
                "result": "I can calculate mathematical expressions and extract keywords. Please include 'calculate' or 'keywords' in your query."
            }

    except Exception as e:

        return {
            "type": "error",
            "result": str(e)
        }
```

## Step 6: Validate the Intelligent Routing Agent

The validation includes:

- Mathematical calculation requests
- Keyword extraction requests
- General conversational queries
- Invalid or empty inputs

Each response is displayed as a structured JSON object containing the response type and the generated result.

```python
# Validation Test Queries

test_queries = [
    "Calculate 20 + 5",
    "Calculate (10 * 4) - 8",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "Extract keywords from Machine Learning improves prediction accuracy",
    "Hello",
    "",
    123
]

for query in test_queries:

    print(f"\nQuery: {query}")

    response = agent(query)

    print(json.dumps(response, indent=4))
```

## Step 7: Interactive Query Interface

The application continuously accepts user queries until the user enters **"exit"**. Every query is processed by the routing agent, and the corresponding response is displayed as a formatted JSON object.

```python
while True:

    user_query = input("\nEnter your query (type 'exit' to quit): ").strip()

    if user_query.lower() == "exit":
        print("Thank you for using the Intelligent Query Routing Agent!")
        break

    response = agent(user_query)
    print(json.dumps(response, indent=4))
```
